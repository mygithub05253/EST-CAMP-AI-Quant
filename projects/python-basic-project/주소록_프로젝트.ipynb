# 프로젝트 개요

- 구현할 기능 (요구사항)
    - 새로운 정보를 주소록에 등록하기
        - 1번 버튼
        - 아이디, 이름, 나이, 키, 몸무게 입력
        - 나이 : 정수
        - 키, 몸무게 : 실수
        - 사용자가 의도치 않은 입력은 하지 않는다고 가정
        - 아이디 중복 입력도 하지 않는다고 가정

    - 기존 주소록 정보를 삭제하기
        - 3번 버튼
        - 삭제할 아이디 입력
        - 입력된 아이디 기반 데이터 삭제

    - 기존 주소록 정보를 수정하기
        - 2번 버튼
        - 수정할 아이디 입력
        - 수정하고 싶은 정보(이름, 나이, 키, 몸무게) 다시 입력
        - 입력된 정보로 기존 내용 덮어쓰기

    - 특정 주소록 정보를 검색하기

    - 전체 주소록 정보를 출력하기
        - 4번 버튼
        - 입력된 모든 주소록을 순차적으로 출력
 
    - 프로그램 종료
        - 0번 버튼

[[id1, 이름, 나이, 키, 몸무게],
[id2, 이름, 나이, 키, 몸무게], ...]

```python
user_lst = []
while True:
    menu = int(input("1 : 추가, 2 : 수정, 3 : 삭제, 4 : 목록, 0 : 종료"))

    if menu == 0: # 종료
        print("프로그램을 종료합니다.")
        break
    
    elif menu == 1: # 추가
        user_id = input("아이디를 입력하세요 : ")
        name = input("이름을 입력하세요 : ")
        age = int(input("나이를 입력하세요 : "))
        height = float(input("키를 입력하세요 : "))
        weight = float(input("몸무게를 입력하세요 : "))
        user_lst.append([user_id, name, age, height, weight])
        print(user_lst)
    
    elif menu == 2: # 수정
        user_mod = input("수정할 아이디를 입력하세요 : ")
        is_found = False
        
        for i in range(len(user_lst)):
            if user_mod == user_lst[i][0]:
                user_lst[i][1] = input("이름을 입력하세요 : ")
                user_lst[i][2] = int(input("나이를 입력하세요 : "))
                user_lst[i][3] = float(input("키를 입력하세요 : "))
                user_lst[i][4] = float(input("몸무게를 입력하세요 : "))
                is_found = True
                break
                
        if not is_found:
            print("수정할 아이디가 존재하지 않습니다.")
    
    elif menu == 3: # 삭제
        user_mod = input("삭제할 아이디를 입력하세요 : ")
        is_found = False
        
        for i in range(len(user_lst)):
            if user_mod == user_lst[i][0]:
                user_lst.remove(user_lst[i])
                is_found = True
                print("해당 아이디를 삭제했습니다.")
                break
                
        if not is_found:
            print("삭제할 아이디가 존재하지 않습니다.")
    
    elif menu == 4: # 목록
        print(user_lst)
    
    else:
        print("잘못된 번호를 입력했습니다.")
        break
        
```

## 주소록 프로젝트 해설

```python
db = [["a", "가나다1", 61, 165.4, 60.2], ["b", "가나다2", 32, 185.4, 80.3]]
while True:
    menu = int(input("1 : 추가, 2 : 수정, 3 : 삭제, 4 : 목록, 0 : 종료"))

    if menu == 0: # 종료
        print("프로그램을 종료합니다.")
        break
    
    elif menu == 1: # 추가
        input_id = input("아이디 입력 : ")
        name = input("이름 입력 : ")
        age = input("나이 입력 : ")
        height = input("키 입력 : ")
        weight = input("몸무게 입력 : ")
        db.append([input_id, name, int(age), float(height), float(weight)])
    
    elif menu == 2: # 수정
        update_id = input("수정할 아이디 입력 : ")
        name = input("이름 : ")
        age = int(input("나이 : "))
        height = float(input("키 : "))
        weight = float(input("몸무게 : "))

        for idx, member in enumerate(db):
            if member[0] == update_id:
                db[idx] = [update_id, name, age, height, weight]
                break
    
    elif menu == 3: # 삭제
        # 삭제할 아이디의 위치로 리스트에서 삭제하기
        delete_id = input("삭제할 아이디 입력 : ")
        for idx, member in enumerate(db):
            if member[0] == delete_id:
                print(idx)
                del db[idx]
                break
    
    elif menu == 4: # 목록
        for i in db:
            print(i)
    
    else:
        print("잘못된 번호를 입력했습니다.")
        break
```
