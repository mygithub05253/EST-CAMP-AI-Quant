# 파일 처리

- 파일 처리를 위해서는 파일 열기(open)를 먼저 해야 함
    - 파일을 열면 파일 읽기(read) 또는 파일 쓰기(write)를 할 수 있음
    - 파일 읽기(읽기) : 파일의 내용을 읽어들이는 것
    - 파일 출력(쓰기) : 파일에 새로운 내용을 추가하거나 새로운 파일을 생성하는 것
- 파일 열기(open)를 수행한 뒤 작업이 마무리를 하면 항상 파일을 닫아줘야 함(close)

## 파일 생성

- 파일 객체 = open(파일이름(경로), 파일 열기 모드)

```python
f = open("./text/new_file.txt", "w")
f.close() # 파일 닫기
```

- 파일 열기 모드
    - r(읽기)
        - 파일을 읽기만 할 때
        - 파일이 없으면 error
    - w(쓰기)
        - 파일에 내용을 쓸 때
        - 파일이 없으면 새로 생성
        - 기존에 파일에 있던 데이터를 완전히 지우고 새로 작성
    - a(추가)
        - 파일의 마지막에 새로운 내용을 추가
        - 파일이 없으면 새로 생성

## 파일 출력

```python
f = open("./text/new_file.txt", "w")

for i in range(1, 11):
    f.write(str(i) + "\n")

f.close()
```

## 파일 입력
- readline()

```python
f = open("./text/new_file.txt", "r")

# 파일의 첫 번째 줄을 읽음
line = f.readline()

print(line)

f.close()
```

```python
# 파일의 모든 줄을 읽고 싶은 경우 반복문 작성
f = open("./text/new_file.txt", "r")

while True:
    line = f.readline()

    if not line:
        break

    # 파일 줄 끝에 \n이 있으므로 빈 줄이 같이 출력됨
    # print(line)

    # 빈 줄을 없애는 방법
    # version1
    # print(line, end= "")

    # version2
    print(line.strip())

f.close()
```

- readlines()
    - 파일의 모든 줄을 읽어서 각각의 줄을 요소로 갖는 리스트가 반환

```python
f = open("./text/new_file.txt", "r")

lines = f.readlines()

for line in lines:
    print(line)

f.close()
```

```python
lines
```

- read()
    - 파일 내용 전체를 문자열로 반환

```python
f = open("./text/new_file.txt", "r")

data = f.read()

print(data)

f.close()
```

```python
data
```

## 파일 수정

```python
f = open("./text/new_file.txt", "a")

for i in range(11, 20):
    f.write(str(i) + "\n")

f.close()
```

```python
f = open("./text/new_file.txt", "r")

data = f.read()

print(data)

f.close()
```

# with문

- 파일 사용이 끝난 다음에는 반드시 close()로 파일을 닫아줘야 함
- 자동으로 파일을 닫아주는 문법이 with문

```python
with open("./text/new_file.txt", "r") as f:
    data = f.read()
    print(data)
```
