# 자료형

- 자료(data) : 프로그램이 처리할 수 있는 모든 것
- 자료형(data type) : 자료를 개발자들이 쉽게 사용할 수 있도록 기능과 역할에 따라서 구분한 종류
  - 문자열(string) : 메일 제목, 메시지 내용 등
    - 예) "안녕하세요", "Hello"

  - 숫자(number) : 물건의 가격, 학생의 성적 등
    - 예) 52, 273, 103.32

  - 불(boolean) : 사용자의 로그인 상태 등
    - 예) True, False

## 숫자 자료형

- 숫자 자료형의 종류
  - 정수형(integer) : 소수점이 없는 숫자
    - 예) 0, 273, -53

  - 실수형(floating point) : 소수점이 있는 숫자
    - 예) 0.0, 52.273, -1.2

- 주의사항
  - 0과 0.0은 같은 크기를 표현하지만 소수점의 여부에 따라 숫자의 자료형이 달라짐

### 정수형(integer)

#### 변수(Variable)

- 메모리에 데이터를 저장하는 이름
- 동적 타이핑(Dynamic Typing) : 타입 변경 가능
- 예) a = 123, b = -178

#### 상수(Constant)

- 값을 한 번 정하면 변경하지 않음
- Python, 언어 차원 상수 키워드 존재하지 않음
- 관례 : 대문자 변수명 사용
- 예) PI = 3.14159

```python
a = 123
b = -178
c = 0

print(a)
print(b)
print(c)
```

```python
# type() : 자료형 확인하기
# type 함수의 괄호 내부에 자료를 넣으면 해당 자료가 어떤 자료형을 가지고 있는지 확인

print(type(a))
print(type(b))
print(type(c))
```

### 실수형(Floating-point)

```python
a = 1.2
b = -3.45

print(a)
print(b)
print(type(a))
print(type(b))
```

#### 실수형의 지수 표현 방식

- e+숫자 혹은 e숫자 : 소수점에서 양의 방향(오른쪽)으로 숫자만큼 이동
    - 예) a = 4.24e4 => 4.24 -> 42400.0

- e-숫자 : 소수점에서 음의 방향(왼쪽)으로 숫자만큼 이동
    - 예) b = 4.24e-4 => 4.24 -> 0.000424

```python
a = 4.24e4
b = 4.24e-4

print(a)
print(b)
print(type(a))
print(type(b))
```

### 사칙연산

```python
a = 3
b = 4

print(a + b) # 합
print(a - b) # 차
print(a * b) # 곱
print(a / b) # 나눗셈
```

#### 연산자의 우선순위

- 숫자를 계산할 때는 곱셈과 나눗셈이 덧셈과 뺄셈보다 우선함
- 같은 우선순위를 가지는 연산자는 왼쪽에서 오른쪽 순서로 계산

```python
print(2 + 2 - 2 * 2 / 2 * 2)
```

```python
# 먼저 연산하고 싶은 부분은 괄호로 표현하기
print((5 + 3) * 2)
print(5 + 3 * 2)
```

```python
# 연산자 우선순위가 확실한 경우에도 명확하게 표현하기 위해 괄호를 사용하기도 함
# 코드가 길어지거나 복잡해지는 경우 헷갈리는 것을 방지하기 위해서 사용하는 것을 권장
print(5 + (3 * 2))
```

#### 문자열 포맷팅(Formatting)

```python
# 첫 번째 방법 : % 연산자 활용
a = 3
b = 4
pi = 3.141592

# 문자열 포맷 코드
print("%d + %d = %d"%(a, b, a + b))

# 문자열 포맷 코드에서 소수점 표현
print("%f"%pi)

# 문자열 포맷 코드에서 소수점 자릿수 정해서 표현
print("%0.2f"%pi)

# 특정 칸에 출력하기 (숫자 자리수만큼의 숫자가 들어온다는 의미)
print("%5d"%(a))

# 특정 칸에 출력하되 비어있는 공간에 숫자 집어넣는 방법 : %(숫자)(숫자자리수)d
print("%05d"%(a))
```

```python
# 두 번째 방법 : format 함수
print("{0} + {1} = {0}".format(a, b, (a + b)))

# format 함수를 이용할 경우 이름으로 값을 넣는 것 가능
print("I ate {num} apples. So I was sick for {day} days.".format(num = 10, day = 3))

# format 함수를 이용해 소수점 표현
print("{0:0.2f}".format(pi))
```

```python
# 세 번째 방법 : f-string
print(f"{a} + {b} = {a + b}")

# f-string을 사용한 소수점 표현
print(f"{pi:0.2f}")
```

## 불 자료형 (Boolean)

- 참(True)과 거짓(False)을 나타내는 자료형
    - 어떤 명제의 결과를 표현
 
- 불린 또는 불리언이라는 발음으로 부름
    - 프로그래밍에서는 짧게 Bool이라고 쓰기도 함

```python
a = True
b = False

print(a)
print(type(a))
print(b)
print(type(b))
```

### 불 연산

#### 비교 연산자

- == : 같다
- != : 다르다
- <, > : 작다, 크다
- <=, >= : 작거나 같다, 크거나 같다

```python
1 == 1
```

```python
1 != 1
```

```python
# 문자열에 비교 연산자를 적용하면 한글은 사전 순서로 앞에 있는 것이 작은 값을 가지게 된다.
print("가방" > "하마")
print("가방" < "하마")
```

#### 논리 연산자

- not : 불을 반대로 전환
- and : 두 값이 모두 참일 때 True를 출력하고, 그 외에는 모두 False 출력
- or : 두 값 중 하나만 참이어도 True를 출력하고, 두 개 모두가 거짓일 때만 False

```python
print(not True)
print(not False)
```

```python
print(True and True)
print(True and False)
print(False and True)
print(False and False)
```

```python
print(True or True)
print(True or False)
print(False or True)
print(False or False)
```
