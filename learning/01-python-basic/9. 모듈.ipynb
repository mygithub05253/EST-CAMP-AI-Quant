## random

- 난수 생성 모듈

```python
import random
```

- randint() : 전달하는 두 인수 사이의 정수를 임의로 생성

```python
random.randint(1, 10)
```

- randrange() : 특정 범위에 속한 정수 중에서 하나를 임의로 생성

```python
random.randrange(10)
```

```python
random.randrange(1, 10)
```

- random()
    - 0 이상 1미만 범위에서 임의의 실수를 생성
    - 0% 이상 100% 미만으로 확률을 처리할 때도 사용

```python
random.random()
```

```python
# 50% 확률로 안녕하세요를 출력하는 코드
if random.random() <= 0.5:
    print("안녕하세요")
```

- choice() : 전달된 시퀀스 자료형에 속한 요소 중에서 하나를 임의로 반환

```python
seasons = ["spring", "summer", "fall", "winter"]

print(random.choice(seasons))
```

- sample(요소, 개수)
    - 전달된 시퀀스 자료형에 속한 요소 중 지정된 개수의 요소를 임의로 반환
    - 반환 결과는 리스트 자료형
    - 중복 없이 선택

```python
random.sample(range(1, 46), 6)
```

```python
sorted(random.sample(range(1, 46), 6))
```

- shuffle()
    - 임의로 섞는 것
    - 전달된 시퀀스 자료형에 속한 요소의 순서를 임의로 조정하여 다시 재배치
    - 실제로 전달된 시퀀스 자료형의 순서가 재배치됨
    - str이나 튜플 자료형을 전달하면 에러 발생

```python
my_list = [1, 2, 3, 4, 5]
random.shuffle(my_list)
```

```python
print(my_list)
```

# time

- 시간 처리와 관련된 모듈

```python
import time
```

- time()
    - 1970년 1월 1일 0시 0분 0초부터 현재까지 경과된 시간을 반환(UNIX OS 배포 날짜 기준)
    - 소수점 이하는 마이크로초(ms)를 의미

```python
time.time()
```

- ctime() : 인수로 전달된 시간을 형식을 갖춰 반환

```python
time.ctime(time.time())
```

- sleep() : 특정 시점동안 일시 정지

```python
# 1초간 일시 정지
time.sleep(1)
```

```python
S = time.time()
time.sleep(1)
print(time.time() - S)
```

# datetime

- 날짜와 시간 데이터를 처리

```python
import datetime
```

- now() : 시스템의 현재 날짜와 시간을 반환

```python
print(datetime.datetime.now())
```

- date() : 특정 날짜를 만들어서 반환

```python
print(datetime.date(2026, 7, 22))
```

- time() : 특정 시간을 만들어서 반환

```python
print(datetime.time(10, 48, 0))
```

- 날짜/시간 관련 필드값

```python
today = datetime.datetime.now()
today
```

```python
today.year
```

```python
today.month
```

```python
today.day
```

```python
today.hour
```

```python
today.minute
```

```python
today.second
```

- timedelta() : 날짜/시간 데이터의 연산을 위해 사용

```python
today = datetime.datetime.now()
yesterday = today - datetime.timedelta(days = 1)
tomorrow = today + datetime.timedelta(days = 1)
```

```python
print(yesterday)
```

```python
print(tomorrow)
```

```python
date1 = datetime.date(2026, 6, 21)
date2 = datetime.date(2026, 5, 1)

date1 - date2
```

```python
# total_seconds() : 특정 기간에 포함된 총 시간을 초로 반환
(date1 - date2).total_seconds()
```

# 라이브러리와 프레임워크

- 라이브러리(library)
    - 정상적인 제어를 하는 모듈

- 프레임워크(framework)
    - 제어 역전이 발생하는 모듈
        - 제어 역전(IoC; Inversion of Control)

## 라이브러리 = 정상적인 제어

- 라이브러리는 모듈의 기능을 개발자가 직접 호출

```python
from math import sin, cos, tan, floor, ceil
```

```python
print(sin(1))
print(cos(1))
print(tan(1))
print(floor(2.5))
print(ceil(2.5))
```

## 프레임워크

- 직접적으로 무언가를 진행하는 코드는 없음

```python
from flask import Flask
```

```python
app = Flask(__name__)

@app.route("/")
def hello():
    return "<h1>Hello World!</h1>"

app.run()
```

- Flask 모듈이 제공하는 명령어를 실행하면 코드를 직접 실행하지 않아도 Flask 모듈 내부에서 서버를 실행한 뒤 지정한 파일을 읽어들여 적절한 상황에 스스로 실행
    - 모듈이 개발자가 작성한 코드를 실행하는 형태의 모듈을 프레임워크라고 부름

- 개발자가 모듈의 함수를 호출하는 것이 일반적인 제어

- 반대로 개발자가 만든 함수를 모듈이 실행하는 것이 제어 역전

- 제어 역전 여부로 라이브러리와 프레임워크를 구분
