# 조건문

- 특정 조건을 만족하는지 여부에 따라 실행하는 코드가 달라야 할 때 사용

```
if 조건식(불 값이 나오는 표현식):
    조건식의 결과가 True일 때 실행문
```

````python
if True:
    print("True입니다")
````

````python
# 조건식이 False인 경우에는 코드 블록에 있는 실행문은 실행되지 않음
if False:
    print("False입니다")
````

## if-else문

- 조건식을 만족하는 경우와 만족하지 않는 경우를 구분하여 코드를 작성할 때 사용
- 조건문이 두 가지로만 구분될 때 if-else문을 사용하면 비교를 한 번만 하기 때문에 if만 사용할 때보다 효율적

```
if 조건식:
    조건식의 결과가 True일 때 실행문
else:
    조건식의 결과가 False일 때 실행문
```

````python
num = 13

if num % 2 == 1:
    num += 1
else:
    num *= 2
````

````python
money = False

if money:
    print("택시를 타고 가라")

else:
    print("걸어 가라")
````

````python
money = 20000
pay = 3000

if money >= pay:
    print("택시를 타고 가라")

else:
    print("걸어 가라")
````

- in, not in
    - x in 반복 가능한 객체
    - x not in 반복 가능한 객체

````python
print(1 in [1, 2, 3])
print(1 not in [1, 2, 3])
print("a" in ("a", "b", "c"))
print("j" in "python")
````

````python
pocket = ["cellphone", "money"]

if "money" in pocket:
    print("택시를 타고 가라")

else:
    print("걸어 가라")
````

if-elif문

- 특정 조건에 따라 3가지 이상으로 구분해야 할 때 사용

```
if 조건식1:
    조건식1의 결과가 True일 때 실행문
elif 조건식2:
    조건식1의 결과가 False이고, 조건식2의 결과가 True일 때 실행문
elif 조건식3:
    조건식1, 2의 결과가 모두 False이고, 조건식3의 결과가 True일 때 실행문
else:
    조건식 1, 2, 3의 결과가 모두 False일 때 실행문
```

````python
pocket = ["cellphone"]
card = True

if "money" in pocket:
    print("돈으로 택시를 타고 가라")

elif card:
    print("카드로 택시를 타고 가라")

else:
    print("걸어 가라")
````

- if 조건문에 불이 아닌 다른 값이 올 때는 자동으로 불로 변환해서 처리
    - 따라서 어떤 값이 True로 변환되고, 어떤 값이 False로 변환되는 지 파악하고 있어야 함

````python
if 0:
    print("참")

else:
    print("거짓")
````

````python
if 1:
    print("참")

else:
    print("거짓")
````

````python
if "Hello":
    print("참")

else:
    print("거짓")
````

### 조건부 표현식

```
<조건문이 참인 경우의 값> if <조건문> else <조건문이 거짓인 경우의 값>
```

````python
score = 60

if score >= 60:
    message = "success"

else:
    message = "failure"

print(message)
````

````python
score = 50
message = "success" if score >= 60 else "failure"
print(message)
````

# while 문

- 특정 조건을 만족하는 동안 반복해서 수행해야 할 때 사용
- 조건이 True인 동안 반복해서 실행
- 반복해야하는 횟수나 값의 범위가 명확하지 않은 경우에 주로 사용

```
while 조건식:
    반복 실행문
```

````python
hits = 0
tree_hp = 10

while hits < tree_hp:
    hits += 1
    print(f"나무를 {hits}번째 찍었습니다")

    if hits == 10:
        print("나무가 넘어갑니다")
````

## while문 중첩

````python
day = 1

while day <= 5:
    hour = 1

    while hour <= 3:
        print(f"{day}일차 {hour}교시입니다.")
        hour += 1

    day += 1
````

````python
print(day, hour)
````

## break

- 반복문을 강제로 종료시킬 때 사용

````python
n = 1

while True:
    print(n)

    if n == 10:
        break

    n += 1
````

````python
print(n)
````

## continue

- 반복문의 시작 지점으로 이동
- 반복에서 제외하거나 생략하고 싶은 코드가 있을 때 사용

````python
a = 0

while a < 10:
    a += 1

    if a % 2 == 0:
        continue

    print(a)
````

# for문

- 값의 범위나 반복 횟수가 정해져 있을 때 주로 사용하는 반복문

- 반복 가능 객체 : 문자열, 리스트, 튜플, range, 세트, 딕셔너리

```
for 변수 in 반복가능객체:
    반복실행문
```
