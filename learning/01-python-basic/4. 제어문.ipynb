# 조건문

- 특정 조건을 만족하는지 여부에 따라 실행하는 코드가 달라야 할 때 사용

```
if 조건식(불 값이 나오는 표현식):
    조건식의 결과가 True일 때 실행문
```

````python
if True:
    print("True입니다")
````

````python
# 조건식이 False인 경우에는 코드 블록에 있는 실행문은 실행되지 않음
if False:
    print("False입니다")
````

## if-else문

- 조건식을 만족하는 경우와 만족하지 않는 경우를 구분하여 코드를 작성할 때 사용
- 조건문이 두 가지로만 구분될 때 if-else문을 사용하면 비교를 한 번만 하기 때문에 if만 사용할 때보다 효율적

```
if 조건식:
    조건식의 결과가 True일 때 실행문
else:
    조건식의 결과가 False일 때 실행문
```

````python
num = 13

if num % 2 == 1:
    num += 1
else:
    num *= 2
````

````python
money = False

if money:
    print("택시를 타고 가라")

else:
    print("걸어 가라")
````

````python
money = 20000
pay = 3000

if money >= pay:
    print("택시를 타고 가라")

else:
    print("걸어 가라")
````

if-elif문

- 특정 조건에 따라 3가지 이상으로 구분해야 할 때 사용

```
if 조건식1:
    조건식1의 결과가 True일 때 실행문
elif 조건식2:
    조건식1의 결과가 False이고, 조건식2의 결과가 True일 때 실행문
elif 조건식3:
    조건식1, 2의 결과가 모두 False이고, 조건식3의 결과가 True일 때 실행문
else:
    조건식 1, 2, 3의 결과가 모두 False일 때 실행문
```

````python
pocket = ["cellphone"]
card = True

if "money" in pocket:
    print("돈으로 택시를 타고 가라")

elif card:
    print("카드로 택시를 타고 가라")

else:
    print("걸어 가라")
````

````python
if 0:
    print("참")

else:
    print("거짓")
````

````python
if 1:
    print("참")

else:
    print("거짓")
````

### 조건부 표현식

```
<조건문이 참인 경우의 값> if <조건문> else <조건문이 거짓인 경우의 값>
```

````python
score = 60

if score >= 60:
    message = "success"

else:
    message = "failure"

print(message)
````

````python
score = 50
message = "success" if score >= 60 else "failure"
print(message)
````

# while 문

- 특정 조건을 만족하는 동안 반복해서 수행해야 할 때 사용
- 조건이 True인 동안 반복해서 실행
- 반복해야하는 횟수나 값의 범위가 명확하지 않은 경우에 주로 사용

```
while 조건식:
    반복 실행문
```

````python
hits = 0
tree_hp = 10

while hits < tree_hp:
    hits += 1
    print(f"나무를 {hits}번째 찍었습니다")

    if hits == 10:
        print("나무가 넘어갑니다")
````

## while문 중첩

````python
day = 1

while day <= 5:
    hour = 1

    while hour <= 3:
        print(f"{day}일차 {hour}교시입니다.")
        hour += 1

    day += 1
````

````python
print(day, hour)
````

## break

- 반복문을 강제로 종료시킬 때 사용

````python
n = 1

while True:
    print(n)

    if n == 10:
        break

    n += 1
````

````python
print(n)
````

## continue

- 반복문의 시작 지점으로 이동
- 반복에서 제외하거나 생략하고 싶은 코드가 있을 때 사용

````python
a = 0

while a < 10:
    a += 1

    if a % 2 == 0:
        continue

    print(a)
````

# for문

- 값의 범위나 반복 횟수가 정해져 있을 때 주로 사용하는 반복문

- 반복 가능 객체 : 문자열, 리스트, 튜플, range, 세트, 딕셔너리

```
for 변수 in 반복가능객체:
    반복실행문
```

````python
test = ["one", "two", "three"]

for i in test:
    print(i)

# 리스트에 있는 요소 하나하나가 i라는 변수에 들어가며, 차례대로 반복
````

````python
# for 반복문은 문자열을 함께 사용할 수도 있음
for char in "안녕하세요":
    print(char)
````

````python
# 5명의 학생이 시험을 보았는데, 시험 점수가 60점을 넘으면 합격이고, 그렇지 않으면 불합격
# 학생들의 시험 점수를 차례대로 검사하는 프로그램
scores = [90, 25, 67, 45, 80]
number = 0

for score in scores:
    number += 1

    if score >= 60:
        print(f"{number}번 학생은 합격입니다.")

    else:
        print(f"{number}번 학생은 불합격입니다.")
````

## for문 중첩

- 일반적으로 n차원 처리를 할 때 사용
    - 리스트가 한 겹으로 감싸지면 1차원 리스트
        - [1, 2, 3]
    - 리스트가 두 겹으로 감싸지면 2차원 리스트
        - [[1, 2, 3], [4, 5, 6]]
    - 위와 같은 n차원 리스트 요소를 모두 확인하려면 반복문을 n번 중첩해야 함

````python
over_list = [
    [1, 2, 3],
    [4, 5, 6, 7],
    [8, 9]
]
````

````python
for items in over_list:
    for item in items:
        print(item)
````

## continue

````python
scores = [90, 25, 67, 45, 80]
number = 0

for score in scores:
    number += 1

    if score < 60:
        continue

    print(f"{number}번 학생은 합격입니다.")
````

## range

- 수열을 자동으로 만들어주는 함수
- range(시작 인덱스, 종료 인덱스, 증감값)

````python
for i in range(10):
    print(i)
````

````python
for i in range(10):
    print(i, end = " ")
````

````python
for i in range(1, 11):
    print(i, end = " ")
````

````python
for i in range(1, 11 ,2):
    print(i, end = " ")
````

````python
# 1부터 10까지의 자연수의 누적합을 구하는 코드
sum = 0
for i in range(1, 11):
    sum += i

print(sum)
````

````python
# 단을 입력하면 해당 구구단을 출력하는 코드
dan = 2

for i in range(1, 10):
    print(f"{dan} * {i} = {dan * i}")
````

````python
# 구구단을 출력하는 코드
dan = 2

for i in range(dan, 10):
    for j in range(1, 10):
        print(f"{i} * {j} = {i * j}")
````

## for 반복문과 딕셔너리 조합

- for 반복문과 딕셔너리를 조합하면 딕셔너리의 키가 변수로 들어감

````python
dict1 = {
    "name" : "건조 망고",
    "type" : "당절임",
    "ingredient" : ["망고", "설탕", "나트륨", "색소"]
}
````

````python
for key in dict1:
    print(key)
    print(dict1[key])
    print("-" * 20)
````

## enumerate 함수

- 리스트의 요소를 반복할 때 요소의 인덱스를 함께 반환
- 현재 인덱스가 몇 번째인지 확인하고 싶은 경우 사용

````python
example_lst = ["요소1", "요소2", "요소3"]

for idx, value in enumerate(example_lst):
    print(f"{idx}번째 요소는 {value}입니다")
````

## 리스트 내포 (List Comprehension)

````python
%%timeit
result1 = []

for num in range(10):
    result1.append(num * 3)

result1
````

````python
%%timeit
result2 = [num * 3 for num in range(10)]

result2
````

````python
result = [num * 3 for num in range(10) if num % 2 == 0]
result
````
