# 컬렉션

- 여러 값을 하나의 이름으로 묶어서 관리하는 자료형

| 컬렉션 | 생성 함수 | 특징 | 예시 |
| :-- | :-- | :-- | :-- |
| 리스트 | list() | 추가, 수정, 삭제가 언제나 가능 | a = [1, 2, 3]|
| 튜플 | tuple() | 생성되고 나면 변경 불가능 | a = (1, 2, 3) |
| 세트 | set() | 중복된 값의 저장 불가능 | a = {1, 2, 3} |
| 딕셔너리 | dict() | 키 + 값으로 관리 | a = {'age':25} |

- 이 중에서 저장된 값들의 순서 있는 컬렉션을 "시퀀스(sequence)"라고 부름
    - 시퀀스 : 리스트, 튜플

## 리스트(list)

- 여러 가지 자료를 저장할 수 있는 자료
    - 지금까지 활용한 숫자나 문자, 수식 등이 개별적인 자료라면, 리스트는 이런 자료들을 모아서 사용할 수 있게 해주는 특별한 형태의 자료
 
- 리스트 생성하는 방법 : 대괄호에 자료를 쉼표로 구분해서 입력하기
    - [요소1, 요소2, 요소3, ...]
    - 대괄호 내부에 넣는 자료를 요소(element)라고 함

```python
str1 = "안녕하세요"
str2 = "!"

print(str1 += str2)
```

```python
# 숫자 만으로 구성된 리스트
list_num = [1, 2, 3, 4]

print(list_num)
```

```python
# 문자열로 구성된 리스트
list_str = ["a", "b", "c"]

print(list_str)
```

```python
# 여러 자료형으로 구성된 리스트
list_mix = [273, 32, 103.1, "문자열", True, False]

print(list_mix)
```

```python
# 리스트 인덱싱, 슬라이싱
print(list_num[0])

print(list_num[-1])
```

```python
# 이중 리스트
li = [1, 2, 3, ["a", "b", "c"]]

print(li)
```

```python
print(li[-1])
```

```python
print(li[-1][0])
```

```python
print(li[:3])
```

## 리스트 연산

- 문자열에 적용할 수 있는 연산자는 리스트에서도 활용될 수 있음

```python
# 리스트 병합
a = [1, 2, 3]
b = [4, 5, 6]

print(a + b)
```

## 리스트 요소 수정, 삭제

- 수정 : 인덱싱 -> 위치에 수정하고 싶은 값 할당

- 삭제
    - 위치 : del, pop()
    - 값 : remove()

```python
li
```

```python
# 요소를 인덱스로 삭제하는 방법
del li[-1] # 슬라이싱 사용 가능
```

```python
print(li)
```

```python
# 요소를 수정하는 방법
li[2]
print(li)

li[2] = 4
print(li)
```

```python
# remove 함수를 이용한 삭제
li.remove(4)

print(li)
```

```python
# remove 함수는 리스트에서 첫 번째로 나오는 특정 요소를 삭제하는 함수
li = [1, 2, 3, 1, 2, 3]

li.remove(3)

print(li)
```

```python
# pop 함수를 이용한 삭제
# 제거할 위치를 입력하지 않으면 기본값(-1)을 적용해서 리스트의 맨 마지막 요소를 반환하고 해당 요소를 삭제
# 사용 형식 : li.pop(삭제할 요소 위치)
print(li.pop())
```

```python
print(li)
```

```python
li.pop(1)

print(li)
```

## 리스트 요소 추가

```python
# 리스트 뒤에 요소를 추가 : append()
print(li)

li.append(5)

print(li)
```

```python
li.append([1, 2, 3])

print(li)
```

```python
# 요소 삽입 : insert(삽입할 위치, 삽입할 값)
# 괄호 안에서 정한 위치에 요소를 삽입
# 해당 위치의 요소는 뒤로 하나씩 밀림
li.insert(0, 4)

print(li)
```

```python
# 리스트 확장 : extend()
# 리스트 뒤에 새로운 리스트의 모든 요소를 추가
# 한 번에 여러 요소를 추가하고 싶을 때 사용
li.extend([4, 5])

print(li)
```

```python
li = [1, 2, 3]
print(li + [4, 5]) # 원본 데이터에 영향을 주지 않음 (비파괴적 명령어)
print(li)
```

```python
li = [1, 2, 3]
li.extend([4, 5]) # 원본 데이터에 직접적인 영향을 줌 (파괴적 명령어)
print(li)
```

- 리스트 연결 연산자와 리스트 확장의 차이
    - 공통점
        - 리스트 연결 연산자와 extend() 함수는 비슷한 형태
            - 리스트와 리스트 연결

    - 차이점
        - 리스트 연결 연산자(+)는 원본 리스트에 영향을 주지 않음
            - 비파괴적 명령어
        - extend() 함수는 원본 리스트에 영향을 줌
            - 파괴적 명령어

    - 기본적으로 자료는 비파괴적으로 사용하는 것이 편리
        - 원본 활용 + 새로운 결과 활용으로 선택지가 넓어지기 때문
        - 따라서 기본 자료형(숫자, 문자열, 불)과 관련된 것들은 비파괴적으로 작동

    - 하지만 리스트는 용량이 매우 커질 수 있음
        - 용량이 얼마나 커질 지 알 수 없기 때문에 원본과 수정 결과를 다 생성하는 것은 위험할 수 있음
        - 이런 경우 원본 데이터를 직접 조작하여 위험 회피

## 리스트 관련 함수

```python
# 리스트 오름차순 정렬 : sort()
li = [1, 4, 3, 2]

li.sort()

print(li)
```

```python
# 리스트 내림차순 정렬 : sort(reverse = True)
li.sort(reverse = True)
print(li)
```

```python
# 리스트 뒤집기 : reverse()
li = ["a", "c", "b", "d"]
li.reverse()
print(li)
```

```python
print(li[::-1])
```

```python
# 리스트의 길이 구하기 : len()
print(len(li))
```

# 튜플

- 리스트와 유사하지만 튜플은 값의 추가, 삭제, 수정이 불가능
    - 프로그램이 실행되는 동안 값이 변하면 안되는 경우 주로 사용

```python
tu = (2, 2, 3, 4, 5)

print(tu)
print(type(tu))
```

```python
del tu[0]
```

```python
tu[0] = 0
```

```python
# 인덱싱, 슬라이싱
print(tu[0])
print(tu[1:])
```

```python
# 튜플 연산
tu1 = (1, 2)
tu2 = (3, 4)

print(tu1 + tu2)
```

```python
print(tu1 * 2)
```

# 딕셔너리

- 자료의 대응 관계를 나타내는 자료형
    - 연관배열(Associative array) 또는 해시(Hash)
 
- 데이터를 조회할 때 key를 통해 value를 구함

- 딕셔너리 선언
    - 중괄호 {}로 선언 (집합과의 공통점)
    - 키 : 값의 형태를 콤마로 연결해서 생성 (집합과의 차이점)
        - 키는 문자열, 숫자, 불 등으로 선언할 수 있지만 일반적으로는 문자열을 주로 사용

```python
dic = {"name" : "andy", "birth" : "1118"}

print(dic)
print(dic["name"])
print(dic["birth"])
```

```python
dic = {"a" : [1, 2, 3]}

print(dic)
print(dic["a"])
```

### 딕셔너리 요소 추가, 수정, 삭제

```python
# 키를 기반으로 값을 입력
dic["name"] = "andy"

print(dic)
```

```python
# 이미 존재하는 키를 지정하고 값을 넣으면 기존의 값을 새로운 값으로 수정
dic["name"] = "파인애플"

print(dic)
```

```python
# 요소 삭제
del dic["name"]

print(dic)
```

- 딕셔너리의 key는 고유한 값
    - 중복을 허용하지 않음
    - 동일한 key가 존재하면 어떤 value를 불러와야할지 알 수 없기 때문

```python
dic = {"name" : "andy", "name" : "파인애플"}

print(dic["name"])
print(dic)
```

```python
dic.keys()
```

- get으로 value 얻기
    - 존재하지 않는 키로 값을 가져오라고 할 경우 오류가 발생함
        - get으로 key가 없을 경우 기본값을 대신 가져오게 할 수 있음

```python
dic.get("foo", "bar")
```

```python
dic.get("birth", "bar")
```

# 세트

- 중복을 허용하지 않음
- 순서가 없음

```python
se = {1, 2, 3, 3, 4}

print(se)
print(type(se))
```

```python
set1 = set("Hello")
print(set1)
```

```python
# 세트 교집합
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}

print(set1 & set2)
```

```python
# 세트 합집합
print(set1 | set2)
```

```python
# 세트 차집합
print(set1 - set2)
print(set2 - set1)
```

```python
# 세트 요소 추가
set1.add(7)

print(set1)
```

```python
# 세트 요소 제거
set1.remove(2)

print(set1)
```
