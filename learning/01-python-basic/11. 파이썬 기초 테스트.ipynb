## 4번

사용자와 컴퓨터가 번갈아 가며 가위, 바위, 보 중 하나를 선택하고, 그에 따른 승패를 결정하는 게임을 작성하시오. 게임의 흐름은 다음과 같음
사용자 입력: 사용자는 "가위", "바위", "보" 중 하나를 입력함. (게임 종료는 0 입력)

컴퓨터 선택: 컴퓨터는 "가위", "바위", "보" 중 무작위로 선택함.

승패 판정: 사용자의 선택과 컴퓨터의 선택에 따라 승패를 판별함.

"가위"는 "보"를 이김.

"바위"는 "가위"를 이김.

"보"는 "바위"를 이김.

같은 선택을 했을 경우 “무승부”.

게임 반복: 사용자가 종료를 원할 때까지 게임은 반복됨.

게임 종료: 사용자가 0을 입력하면 게임이 종료되고, 총 게임 횟수와 사용자 기준의 전적(승/패/무)을 출력.

```python
import random

user_win = 0
computer_win = 0
nobody_win = 0
total_game = 0

while True:
    user_pick = input("가위 / 바위 / 보 중 하나를 입력하세요. 게임 종료는 0을 입력하세요 : ")
    choices = ["가위", "바위", "보"]

    computer = random.choice(choices)

    if user_pick == computer:
        print("무승부")
        nobody_win += 1
        total_game += 1

    elif (user_pick == "가위" and computer == "보") or (user_pick == "바위" and computer == "가위") or (user_pick == "보" and computer == "바위"):
        print("사용자가 이기고 컴퓨터가 졌습니다.")
        user_win += 1
        total_game += 1

    elif (user_pick == "가위" and computer == "바위") or (user_pick == "바위" and computer == "보") or (user_pick == "보" and computer == "가위"):
        print("사용자가 지고 컴퓨터가 이겼습니다.")
        computer_win += 1
        total_game += 1

    elif user_pick == "0":
        print(f"총 게임 횟수 : {total_game}")
        print(f"사용자 기준의 전적(승/패/무) : {user_win}/{computer_win}/{nobody_win}")
        break
```
