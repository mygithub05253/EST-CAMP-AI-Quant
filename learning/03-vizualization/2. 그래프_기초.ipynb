# 막대 그래프

# 산점도

- 서로 다른 두 연속형 변수 사이의 관계를 나타내며 연속형 변수의 상관성을 확인할 때 산점도 그래프를 사용
    - 예) 나이와 소득에 대한 상호 관련성 파악
 
- 산점도에 표시되는 각 점들은 자료의 관측값을 의미하고, 각 점의 위치는 관측값이 가지는 x축, y축 변수의 값으로 결정됨

```python
N = 50
x = np.random.rand(N)
y = np.random.rand(N)
colors = np.random.rand(N)
area = (30 * np.random.rand(N)) ** 2

plt.scatter(x, y, s = area, c = colors, alpha = 0.5)

plt.show()
```

# 파이차트

```python
# 시작은 3시에서 시작하며 반시계 방향으로 진행
ratio = [34, 32, 16, 18]
labels = ["Apple", "Banana", "Melon", "Grapes"]

plt.pie(ratio, labels = labels, autopct = "%.1f%%")

plt.show()
```

```python
# 비율을 100으로 꼭 맞추지 않아도 알아서 100으로 맞춰서 파이 차트를 반환해줌
ratio = [340, 32, 16, 18]
labels = ["Apple", "Banana", "Melon", "Grapes"]

plt.pie(ratio, labels = labels, autopct = "%.1f%%")

plt.show()
```

```python
# 시작하는 위치 선정 : startangle
ratio = [34, 32, 16, 18]
labels = ["Apple", "Banana", "Melon", "Grapes"]

plt.pie(ratio, labels = labels, autopct = "%.1f%%", startangle = 90)

plt.show()
```

```python
ratio = [34, 32, 16, 18]
labels = ["Apple", "Banana", "Melon", "Grapes"]

plt.pie(ratio, labels = labels, autopct = "%.1f%%", startangle = 260)

plt.show()
```

```python
# 시계 방향으로 진행하도록 하는 방법 : counterclock = False
ratio = [34, 32, 16, 18]
labels = ["Apple", "Banana", "Melon", "Grapes"]

plt.pie(ratio, labels = labels, autopct = "%.1f%%", startangle = 260, counterclock = False)

plt.show()
```

```python
# 파이 조각을 중앙을 기준으로 띄우는 방법 : explode 옵션
ratio = [34, 32, 16, 18]
labels = ["Apple", "Banana", "Melon", "Grapes"]
explode = [0, 0.1, 0, 0.1]

plt.pie(ratio, labels = labels, autopct = "%.1f%%", startangle = 260, counterclock = False, explode = explode)

plt.show()
```

```python
ratio = [34, 32, 16, 18]
labels = ["Apple", "Banana", "Melon", "Grapes"]
explode = [0.05, 0.05, 0.05, 0.05]

plt.pie(ratio, labels = labels, autopct = "%.1f%%", startangle = 260, counterclock = False, explode = explode)

plt.show()
```

```python
# 음영을 넣는 방법 : shadow = True
ratio = [34, 32, 16, 18]
labels = ["Apple", "Banana", "Melon", "Grapes"]
explode = [0.05, 0.05, 0.05, 0.05]

plt.pie(ratio, labels = labels, autopct = "%.1f%%", startangle = 260, counterclock = False, explode = explode, shadow = True)

plt.show()
```

### 건강검진 데이터 남성과 여성의 허리둘레 히스토그램

```python
health = pd.read_excel("./data/xlsx/health_screenings_2020_1000ea.xlsx")
```

```python
health.head()
```

```python
# 보기 불편한 데이터 전처리 작업 수행
# 다만 실습을 위해서 임의로 변경하는 것이지 실제 프로젝트나 실무에서는 임의로 변경해서는 안됩니다!
health["gender"] = health["gender"].map(lambda x : "Male" if x == 1 else "Female")
health["drinking"] = health["drinking"].map(lambda x : "Non-drinking" if x == 0 else "Drinking")
health["smoking"] = health["smoking"].map(lambda x : "Smoking" if x == 3 else "Non-smoking")
```

```python
health.head()
```

```python
male_data = health.loc[health["gender"] == "Male", ["gender", "weight", "waist", "drinking", "smoking"]]
female_data = health.loc[health["gender"] == "Female", ["gender", "weight", "waist", "drinking", "smoking"]]
```

```python
plt.figure(figsize = (10, 6))

sns.histplot(male_data["waist"], bins = 7, alpha = 0.5, label = "Male", kde = True)
sns.histplot(female_data["waist"], bins = 7, alpha = 0.5, label = "Female", kde = True)

plt.xlabel("waist")
plt.ylabel("frequency")
plt.title("waist seaborn histogram with kde")
plt.legend()
plt.grid()

plt.show()
```

```python
plt.figure(figsize = (10, 6))

sns.histplot(male_data["waist"], bins = 7, alpha = 0.5, label = "Male", kde = True)
sns.histplot(female_data["waist"], bins = 7, alpha = 0.5, label = "Female", kde = True, color = "r")

plt.xlabel("waist")
plt.ylabel("frequency")
plt.title("waist seaborn histogram with kde")
plt.legend()
plt.grid()

plt.show()
```

```python
plt.figure(figsize = (10, 6))

sns.histplot(male_data["waist"], bins = 7, alpha = 0.5, label = "Male", kde = True)
sns.histplot(female_data["waist"], bins = 7, alpha = 0.5, label = "Female", kde = True, color = "r")

plt.xlabel("waist")
plt.ylabel("frequency")
plt.title("waist seaborn histogram with kde")
plt.legend()
plt.grid()
plt.xlim(40, 140)

plt.show()
```

## 산점도와 히스토그램 한 번에 그리기

```python
joint = sns.jointplot(x = "total_bill", y = "tip", data = tips)

joint.set_axis_labels(xlabel = "지불금액", ylabel = "팁")

joint.fig.suptitle("지불금액과 팁 jointplot")

plt.show()
```

```python
joint = sns.jointplot(x = "total_bill", y = "tip", data = tips)

joint.set_axis_labels(xlabel = "지불금액", ylabel = "팁")

joint.fig.suptitle("지불금액과 팁 jointplot", y = 1.03) # 제목과 그래프 사이의 거리 띄우는 옵션

plt.show()
```

- 산점도는 점이 겹칠 경우 구분하기 어려움 -> 육각 그래프

- 육각 그래프 : 2차원 표면에 육각형으로 데이터를 쌓아 표현, 특정 데이터의 개수가 많아지면 진한 색으로 표현

```python
joint = sns.jointplot(x = "total_bill", y = "tip", data = tips, kind = "hex") # 육각 그래프 옵션

joint.set_axis_labels(xlabel = "지불금액", ylabel = "팁")

joint.fig.suptitle("지불금액과 팁 jointplot", y = 1.03)

plt.show()
```
