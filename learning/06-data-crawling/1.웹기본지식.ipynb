# 인코딩

## ASCII

- 한글이 포함된 엑셀이나 csv 파일을 불러올 때, 또는 한글로 된 데이터를 크롤링하면 오류가 뜨거나 읽을 수 없는 문자로 나타나는 경우가 있음
    - 한글 인코딩(encoding) 때문에 발생하는 문제
 
- 인코딩 : 사람이 사용하는 언어를 컴퓨터가 사용하는 0과 1로 변환하는 과정
    - 디코딩 : 인코딩의 반대 과정
 
- 사람과 컴퓨터 간의 언어를 번역하기 위해 최초로 사용된 방식이 아스키(American Standard Code for Information Interchange, ASCII)
    - 0부터 127까지 128개 바이트에 알파벳과 숫자, 특수문자의 값을 부여하고, 문자가 입력되면 이에 대응되는 바이트가 저장
    - 하지만 영어 알파벳이 아닌 다른 문자 표현에 한계가 있으며, 이를 보완하기 위한 방법이 필요해짐

## 한글 인코딩 방식의 종류

- 한글을 인코딩하는 데 쓰이는 방식은 EUC-KR, CP949, UTF-8

- 한글을 인코딩 하는 방법
    1. 완성형
        - 완성된 문자 자체에 해당하는 코드를 부여
        - 예) "퀀" 을 나타내는 코드, "트" 를 나타내는 코드
        - 대표적인 방법 : EUC-KR, CP949
            - CP949는 EUC-KR에 비해 표현할 수 있는 문자의 개수가 많아졌으며 윈도우의 기본 인코딩 방식
          
    2. 조합형
        - ㅋ, ㅜ, ㅓ, ㄴ, ㅌ, ㅡ 각각에 해당하는 코드를 부여하고 조합
        - 대표적인 방법 : UTF-8
            - 자음과 모음 각각에 코드를 부여한 후 조합해 한글을 나타냄
            - 조합형은 한글뿐만 아니라 다양한 언어에 적용할 수 있어 웹페이지의 대부분이 UTF-8로 만들어지고 있음

# 웹 동작 방식

- 클라이언트
    - 데스크톱이나 휴대폰과 같은 장치와 크롬이나 파이어폭스와 같은 소프트웨어를 의미
 
- 서버
    - 웹사이트와 앱을 저장하는 컴퓨터를 의미
 
- 클라이언트가 정보를 요구하는 과정을 요청(request)이라고 하며, 서버가 해당 정보를 제공하는 과정을 응답(response)라고 함
    - 클라이언트와 서버가 연결되어 있지 않다면 둘 사이에 정보를 주고받을 수 없으며, 이를 연결하는 공간이 인터넷
 
- URL
    - 각 서버의 고유 주소
 
- 예) 네이버에서 경제 기사를 클릭하는 경우
    - 클라이언트 : 사용자
    - 서버 : 네이버
    - URL : www.naver.com
    - 요청 : 경제 기사 클릭(경제 기사 정보 요청)
    - 응답 : 클릭 후 해당 페이지를 보여 주는 과정

## HTTP

- 클라이언트가 각기 다른 방법으로 데이터를 요청한다면 서버가 요청을 알아듣지 못하기 때문에 데이터 요청은 규정된 약속이나 표준에 맞추어 요청해야 함
    - 이러한 약속을 HTTP(HyperText Transfer Protocol)라고 함
 
- HTTP 요청 방식(HTTP request method)
    - 클라이언트가 서버에게 요청의 목적이나 종류를 알리는 방법
 
    - HTTP 요청 방식의 종류
        - GET : 특정 정보 조회
        - POST : 새로운 정보 등록
        - PUT : 기존 특정 정보 갱신
        - DELETE : 기존 특정 정보 삭제

# HTML과 CSS

- 클라이언트와 서버가 데이털르 주고받을 때는 디자인이 필요하지 않음

- 그러나 응답받은 정보를 사람이 확인하려면 보기 편한 방식으로 바꾸어 줘야 함
    - 웹 페이지의 제목, 단락, 목록 등 레이아웃을 잡아 주는 데 사용되는 대표적인 마크업 언어가 HTML(HyperText Markup Language)
 
- HTML을 통해 잡힌 뼈대에 글자의 색상이나 폰트, 배경색, 배치 등 화면을 꾸며주는 역할을 하는 것이 CSS(Cascading Style Sheets)

- 크롤링하고자 하는 데이터가 웹 페이지의 어떤 태그 내에 위치하고 있는지, 어떻게 크롤링하면 될지 파악하기 위해서는 HTML과 CSS에 대한 기본적인 지식이 필요

- HTML, CSS 실습 페이지
    - https://www.w3schools.com/html/tryit.asp?filename=tryhtml_intro

## HTML 기본 구조

- HTML은 메타 데이터를 나타내는 head와 본문을 나타내는 body로 나뉨

## 태그와 속성

- HTML 코드는 태그와 속성, 내용으로 이루어져 있음

- 태그 : 꺾쇠(<>)로 감싸져 있는 부분, 태그가 있으면 이를 닫는 태그인 </> 가 쌍으로 있어야 함

- 속성 : 해당 태그에 대한 추가적인 정보를 제공해 주는 것

- 내용 : 웹페이지에서 눈으로 볼 수 있는 텍스트 부분

## h태그와 p태그

- h태그 : 폰트의 크기를 나타내는 태그

- p태그 : 문단을 나타내는 태그

```html
<html>
    <body>
        <h1>Page heading: size 1</h1>
        <h2>Page heading: size 2</h2>
        <h3>Page heading: size 3</h3>

        <p>Quant Portfolio</p>
        <p>By Jun</p>
    </body>
</html>
```

- h태그의 숫자가 작을수록 텍스트 크기는 커지며, 숫자는 1에서 6까지 지원됨

## 리스트를 나타내는 ul태그, ol태그

- 리스트(글머리 기호)를 만들 때 사용

- ul(unordered list) : 순서가 없는 리스트

- ol(ordered list) : 순서가 있는 리스트

```html
<html>
    <body>
        <h2>Unordered List</h2>
        <ul>
            <li>List 1</li>
            <li>List 2</li>
            <li>List 3</li>
        </ul>

        <h2>Ordered List</h2>
        <ol>
            <li>List 1</li>
            <li>List 2</li>
            <li>List 3</li>
        </ol>
    </body>
</html>
```

## table 태그

- 표를 만드는 태그

```html
<html>
    <body>
        <h2>Sample Table</h2>

        <table>
            <tr>
                <th>Column 1</th>
                <th>Column 2</th>
                <th>Column 3</th>
            </tr>
            <tr>
                <td>1</td>
                <td>2</td>
                <td>3</td>
            </tr>
            <tr>
                <td>A</td>
                <td>B</td>
                <td>C</td>
            </tr>
            <tr>
                <td>a</td>
                <td>b</td>
                <td>c</td>
            </tr>
        </table>
    </body>
</html>
```

- table 태그 내의 tr 태그는 각 행을 의미

- 각 셀의 구분은 th 또는 td 태그를 통해 구분

## a 태그와 img 태그 및 속성

- 속성과 결합해 사용

- a태그는 href 속성과 결합해 다른 페이지의 링크를 걸 수 있음

- img 태그는 src 속성과 결합해 이미지를 불러옴

```html
<html>
    <body>
        <h2>a tag & href attribute</h2>
        <p>HTML links are defined with the a tag.
        The link address is specified in the href attribute:</p>

        <a href="https://blog.naver.com/leebisu">Henry's Quantopia</a>

        <h2>img tag & src attribute</h2>
        <p>HTML images are defined with the img tag,
        and the filename of the image source is
        specified in the src attribute:</p>

        <img src="https://www.python.org/static/img/python-logo.png", width="200", height="100">
    </body>
</html>
```

## div 태그

- 화면의 전체적인 틀을 만들 때 주로 사용하는 태그

- 단독으로도 사용될 수 있으며, 꾸밈을 담당하는 style 속성과 결합되어 사용되기도 함

```html
<html>
    <body>
        <div style="background-color:black;color:white">
            <h5>First Div</h5>
            <p>Black background, White Color</p>
        </div>

        <div style="background-color:yellow;color:red">
            <h5>Second Div</h5>
            <p>Yellow background, Red Color</p>
        </div>

        <div style="background-color:blue;color:grey">
            <h5>Third Div</h5>
            <p>Blue background, Grey Color</p>
        </div>
    </body>
</html>
```

## CSS

- 웹 페이지를 꾸며주는 역할

- head 태그에서 각 태그에 CSS 효과를 입력하면 본문의 모든 해당 태그에 CSS 효과가 적용됨
    - 이처럼 웹 페이지를 꾸미기 위해 특정 요소에 접근하는 것을 셀렉터(selector)라고 함
 
```html
<html>
    <head>
        <style>
            body {background-color: powderblue;}
            h4 {color: blue;}
        </style>
    </head>
    <body>
        <h4>This is a heading</h4>
        <p>This is a first paragraph.</p>
        <p>This is a second paragraph.</p>
    </body>
</html>
```

## 클래스와 id

- 클래스 속성을 이용하면 특정 이름을 가진 클래스에 동일한 효과를 적용할 수 있음

```html
<html>
    <style>
        .language {
            background-color: tomato;
            color: white;
            padding: 10px;
        }
        .desc {
            background-color: moccasin;
            color: black;
            padding: 10px;
        }
    </style>

    <div>
        <h2 class="language">Python</h2>
        <p class="desc">Python is a high-level, general-purpose programming language.</p>
    </div>

    <div>
        <h2>SQL</h2>
        <p>SQL is a domain-specific language used in programming and designed for managing data held in a RDBMS, or for stream processing in a RDBMS.</p>
    </div>

    <div>
        <h2 class="language">R</h2>
        <p class="desc">R is a free software environment for statistical computing and graphics</p>
    </div>
</html>
```

- 셀렉터를 클래스에 적용할 때는 클래스명 앞에 마침표(.)를 붙여 표현

- id 또한 비슷한 역할을 함
    - 다만 HTML 내에서 클래스는 여러 개가 정의될 수 있지만, id는 단 하나만 사용
 
```html
<html>
    <head>
        <style>
            #myHeader {
                background-color: lightblue;
                color: black;
                padding: 15px;
                text-align: center;
            }
        </style>
    </head>
    <body>
        <h1 id="myHeader">My Header</h1>
    </body>
</html>
```

- 셀렉터를 id에 적용할 때는 id명 앞에 샵(#)을 붙여 표현

- 페이지에서 한 번만 사용된다는 점을 제외하면 클래스와 사용 방법은 거의 동일

- 클래스나 id값을 통해 원하는 내용을 크롤링하는 경우가 많음
