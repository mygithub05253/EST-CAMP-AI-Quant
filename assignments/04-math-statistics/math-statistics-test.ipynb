# 5번

배열 arr이 아래와 같을 때, 아래의 지시사항에 따라 데이터 전처리를 수행하는 코드를 각각 작성하시오

arr = np.arange(1, 9).reshape(4, 2)

가. 각 열의 값이 최솟값이 0, 최댓값이 1이 될 수 있도록 정규화를 하려고 한다

정규화의 공식은 ((각 요소 값 - 해당 열의 최솟값) / (해당 열의 최댓값 - 해당 열의 최솟값))

나. 각 행 내에서 해당 요소가 차지하는 비율로 값을 수정하려고 한다
정규화의 공식은 (각 요소 값 / 해당 행의 합)

```python
arr = np.arange(1, 9).reshape(4, 2)

print(arr)
```

```python
# 가. 각 열의 값이 최솟값이 0, 최댓값이 1이 될 수 있도록 정규화를 하려고 한다
# 정규화의 공식은 ((각 요소 값 - 해당 열의 최솟값) / (해당 열의 최댓값 - 해당 열의 최솟값))
arr = np.arange(1, 9).reshape(4, 2)

col_min = arr.min(axis = 0)
col_max = arr.max(axis = 0)

print(col_min)
print(col_max)

result = (arr - col_min) / (col_max - col_min)

print(result)
```

```python
# 나. 각 행 내에서 해당 요소가 차지하는 비율로 값을 수정하려고 한다 
# 정규화의 공식은 (각 요소 값 / 해당 행의 합)
arr = np.arange(1, 9).reshape(4, 2)

row_sum = arr.sum(axis = 1)
print(row_sum)

row_sum = row_sum.reshape(-1, 1)
print(row_sum)

result = arr / row_sum

print(result)
```

# 7번

남성 700명과 여성 500명이 있다.

남성 중 기혼자의 비율은 30% 이고, 여성 중 기혼자의 비율은 20% 이다.

남성과 여성의 성별이 혼인 여부에 영향을 주는지 확인하려 한다.

적절한 통계검정을 진행하고 검증된 p_value 결과를 바탕으로 성별에 따른 혼인여부의 분포에 통계적으로 유의미한 차이가 있는지를 답안으로 작성하시오

단, 유의수준은 0.05를 기준으로 한다.

```python
from scipy.stats import chi2_contingency
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# windows용 한글 폰트 오류 해결
from matplotlib import font_manager, rc
font_path = 'C:/Windows/Fonts/malgun.ttf'
font_name = font_manager.FontProperties(fname = font_path).get_name()
rc('font', family=font_name)
```

```python
male_gender = ["남성"] * 700
male_marriage = ["기혼"] * 210 + ["미혼"] * 490
df_male = pd.DataFrame({"성별": male_gender, "혼인여부": male_marriage})

female_gender = ["여성"] * 500
female_marriage = ["기혼"] * 100 + ["미혼"] * 400
df_female = pd.DataFrame({"성별": female_gender, "혼인여부": female_marriage})

df = pd.concat([df_male, df_female], ignore_index=True)
```

```python
df.head()
```

```python
df.tail()
```

```python
cross = pd.crosstab(df["성별"], df["혼인여부"])
print(cross)
```

```python
cross.plot(kind = "bar", figsize = (10, 5))
plt.grid()

plt.show()
```

```python
chi2_contingency(cross, correction = True)
```

```python
chiresult = chi2_contingency(cross)
```

```python
print(chiresult[0]) # 카이제곱 통계량
```

```python
print(chiresult[1]) # p-value
```

결론
귀무가설 : 성별이 혼인 여부에 영향을 주지 않는다
대립가설 : 성별이 혼인 여부에 영향을 준다

유의수준 0.05 기준 하에서, 검정 결과 계산된 p-value 값은 0.000126으로 유의수준보다 매우 작으므로 귀무가설을 기각하고 대립가설을 채택

성별에 따른 혼인 여부의 분포에는 통계적으로 유의미한 차이가 존재하며, 성별과 혼인 여부는 서로 영향을 준다고 할 수 있다.
