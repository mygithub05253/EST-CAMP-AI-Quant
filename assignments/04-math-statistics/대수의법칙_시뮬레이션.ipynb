```python
import numpy as np
from scipy.stats import gmean, hmean

# 2. 10명 몸무게 표본 데이터(kg)
weights = np.array([60, 62, 65, 68, 70, 72, 75, 78, 80, 85])

# 3. 산술평균 · 기하평균 · 조화평균
arithmetic_mean = np.mean(weights)
geometric_mean = gmean(weights)
harmonic_mean = hmean(weights)

# 4. 결과 출력
print("몸무게 데이터:", weights)
print(f"산술평균: {arithmetic_mean:.2f} kg")
print(f"기하평균: {geometric_mean:.2f} kg")
print(f"조화평균: {harmonic_mean:.2f} kg")
```

```python
import numpy as np
from scipy.stats import randint
import matplotlib.pyplot as plt

trials_list      = [10, 100, 1000, 10000, 100000]
theoretical_prob = 6 / 36
empirical_probs  = []

for n in trials_list:
    die1 = randint.rvs(1, 7, size=n)
    die2 = randint.rvs(1, 7, size=n)

    sums           = die1 + die2
    count_event    = np.sum(sums == 7)
    empirical_prob = count_event / n
    empirical_probs.append(empirical_prob)

    print(f"시행 횟수: {n}")
    print(f"합이 7인 횟수: {count_event}")
    print(f"통계적 확률: {empirical_prob:.6f}")
    print(f"수학적 확률: {theoretical_prob:.6f}")
    print("-" * 30)

plt.figure(figsize=(8, 5))
plt.plot(trials_list, empirical_probs, marker="o", label="empirical")
plt.axhline(theoretical_prob, color="red", linestyle="--", label="theoretical=1/6")
plt.xscale("log")
plt.xlabel("number of trials")
plt.show()
```
