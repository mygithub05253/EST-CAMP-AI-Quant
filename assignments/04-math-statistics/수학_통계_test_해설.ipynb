11. 배열 arr이 아래와 같을 때, 아래의 지시사항에 따라 데이터 전처리를 수행하는 코드를 각각 작성하시오

- arr = np.arange(1, 9).reshape(4, 2)

- 가. 각 열의 값이 최솟값이 0, 최댓값이 1이 될 수 있도록 정규화를 하려고 한다

- 정규화의 공식은 ((각 요소 값 - 해당 열의 최솟값) / (해당 열의 최댓값 - 해당 열의 최솟값))

- 나. 각 행 내에서 해당 요소가 차지하는 비율로 값을 수정하려고 한다
- 정규화의 공식은 (각 요소 값 / 해당 행의 합)

```python
arr = np.arange(1, 9).reshape(4, 2)
```

```python
arr
```

```python
arr.min(axis = 0)
```

```python
arr - arr.min(axis = 0)
```

```python
((arr - arr.min(axis = 0)) / (arr.max(axis = 0) - arr.min(axis = 0)))
```

```python
# 정규화의 공식은 (각 요소 값 / 해당 행의 합)
np.sum(arr, axis = 1)
```

```python
arr
```

```python
np.sum(arr, axis = 1).reshape(-1, 1)
```

```python
arr / np.sum(arr, axis = 1).reshape(-1, 1)
```

13. 남성 700명과 여성 500명이 있다.

- 남성 중 기혼자의 비율은 30% 이고, 여성 중 기혼자의 비율은 20% 이다.

- 남성과 여성의 성별이 혼인 여부에 영향을 주는지 확인하려 한다.

- 적절한 통계검정을 진행하고 검증된 p_value 결과를 바탕으로 성별에 따른 혼인여부의 분포에 통계적으로 유의미한 차이가 있는지를 답안으로 작성하시오

- 단, 유의수준은 0.05를 기준으로 한다.

```python
man = 700
woman = 500

# 기혼자 수
man_married = int(man * 0.3)
woman_married = int(woman * 0.2)

# 미혼자 수
man_single = man - man_married
woman_single = woman - woman_married

data = np.array([[man_married, man_single],
                 [woman_married, woman_single]])
data
```

```python
chi2_contingency(data)
```
