# 제어문 연습문제

## 1번

1부터 1000까지의 자연수 중 3의 배수들의 합을 구하는 코드 작성하기

````python
sum = 0

for i in range(1, 1001):
    if i % 3 == 0:
        sum += i

print(sum)
````

## 2번

반복문 활용하여 계단식으로 별 출력하기

````python
for i in range(1, 6):
    print("*" * i)
````

````python
for i in range(5):
    for j in range(5):
        if j <= i:
            print("*", end = "")
    print()
````

## 3번

화폐매수 구하기 \
예) 156780원이 입력 된다면 \
50000 - 3 \
10000 - 0 \
5000 - 1 \
1000 - 1 \
500 - 1 \
100 - 2 \
50 - 1 \
10 - 3 \
으로 각 화폐 단위별로 필요한 매수를 출력

````python
def count_money(money):
    money_lsts = [50000, 10000, 5000, 1000, 500, 100, 50, 10]
    
    for money_lst in money_lsts:
        count = money // money_lst
        money %= money_lst
        print(f"{money_lst} - {count}")
````

````python
count_money(156780)
````

# 클래스 연습문제

## 1번

아래의 코드가 정상적으로 실행되도록 클래스를 작성하기

```
x = Knight(health = 542.4, mana = 210.3, armor = 38)

print(x.health, x.mana, x.armor)

x.slash() # "베기"라는 문자열이 출력되는 메서드
```

````python
class Knight():
    def __init__(self, health, mana, armor):
        self.health = health
        self.mana = mana
        self.armor = armor

    def slash(self):
        print("베기")
````

````python
x = Knight(health = 542.4, mana = 210.3, armor = 38)

print(x.health, x.mana, x.armor)

x.slash() # "베기"라는 문자열이 출력되는 메서드
````
