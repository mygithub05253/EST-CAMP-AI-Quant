# 함수 연습문제

## 1번

몫과 나머지를 구하는 함수를 작성하기
- 예) a = 10, b = 3
- 반환값 : 3(몫), 1(나머지)

```python
def calc(a, b):
    print(f"몫 : {a} / {b} = {int(a / b)}")
    print(f"나머지 : {a} % {b} = {a % b}")
```

```python
calc(10, 3)
```

```python
def calc(a, b):
    return int(a / b), a % b
```

```python
calc(10, 3)
```

### 1번 해설

```python
def get_divmod(a, b):
    return a // b, a % b
```

## 2번

주어진 자연수가 홀수인지 짝수인지 판별해주는 함수를 작성하기
- 함수명 : is_odd
- 반환값 : 입력된 자연수가 홀수이면 True, 짝수이면 False 반환

```python
def is_odd(num):
    if num % 2 == 1:
        return True
    else:
        return False
```

```python
is_odd(3)
```

```python
is_odd(4)
```

## 3번

커피자판기 코드 만들기

- 커피자판기에 돈과 주문할 커피를 전달
- 주문할 수 있는 커피의 종류와 가격은 다음과 같음
    - 아메리카노 : 1000원
    - 카페라떼 : 1500원
    - 카푸치노 : 2000원

- 없는 커피를 주문할 경우 입력한 돈을 그대로 반환
- 구매 금액이 부족하면 입력한 돈을 그대로 반환
- 정상 주문이면 주문한 커피와 잔돈을 반환

- 예) 커피자판기("마키아또", 15000) -> 15000, 커피자판기("아메리카노", 500) -> 500, 커피자판기("카페라떼", 2000) -> "카페라떼", 500

```python
def coffee_machine(coffee, money):
    coffee_lst = { "아메리카노" : 1000, "카페라떼" : 1500, "카푸치노" : 2000 }

    if coffee in coffee_lst:
        if money < coffee_lst[coffee]:
            return money
        else:
            return coffee, money - coffee_lst[coffee]
    else:
        return money
```

```python
coffee_machine("아메리카노", 500)
```

```python
coffee_machine("마키아또", 15000)
```

```python
coffee_machine("카페라떼", 2000)
```

### 3번 해설

```python
def coffee_machine(pick, money):
    menu = {
        "아메리카노" : 1000,
        "카페라떼" : 1500,
        "카푸치노" : 2000
    }

    if pick not in menu:
        return money
    elif menu[pick] > money:
        return money
    else:
        return pick, money - menu[pick]
```

```python
coffee_machine("아메리카노", 500)
```

```python
coffee_machine("마키아또", 15000)
```

```python
coffee_machine("카페라떼", 2000)
```
