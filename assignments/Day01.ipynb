## 2번

주민등록번호에서 생년월일, 성별 추출하여 출력하기

예) \
hong "881120-1068234" \
hong에서 생년월일 부분 추출하여 출력하는 코드 \
hong에서 성별 부분 추출하여 출력하는 코드

## 3번

비밀번호 만들기

사이트별로 아래의 규칙에 따라 비밀번호를 생성하는 코드 작성하기 \
사이트 주소 예시) "http://naver.com/", "http://daum.net/", "http://google.com/", "http://youtube.com"

규칙
1. http:// 부분은 제외
2. 처음 만나는 점(.) 이후 부분도 제외
3. 남은 글자 중 처음 세 자리 + 글자 개수 + 글자 내 "e"의 개수 + "!"로 비밀번호를 구성
예시) 규칙에 따라 "http://naver.com/"의 비밀번호를 생성하면 nav51!가 됨

```python
site = "http://naver.com/"

site_replace = site.replace(site[0:7], "")
print(site_replace)
```

```python
site_replace2 = site_replace.replace(site_replace[5:], "")
print(site_replace2)
```

```python
print("{a}{b}{c}{d}".format(a = site_replace2[:3], b = len(site_replace2), c = site_replace2.count("e"), d = "!"))
```
