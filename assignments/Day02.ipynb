# 3번
학생의 학점을 입력받아 아래의 조건에 따라 적절한 문장을 출력하기 \
(학점은 0.0 ~ 4.5 범위의 임의의 실수 할당) \

학점이 4.2 이상이면 "교수님의 사랑"
학점이 2.8 이상이고, 4.2미만이면 "일반인"
학점이 2.8 미만이면 "오락문화의 선구자" 

```python
score = 4.3

if score >= 4.2:
    print("교수님의 사랑")
elif score >= 2.8 and score < 4.2:
    print("일반인")
else:
    print("오락문화의 선구자")
```
